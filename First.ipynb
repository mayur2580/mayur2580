### Question:
Write a Python program to:

1. Tokenize the given paragraph
2. Remove stopwords
3. Convert all words to lowercase and remove punctuation
4. Display the cleaned tokens and frequency of each word
5. Count the frequency of words

```python
text = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence.
It helps computers understand, interpret, and respond to human language.
NLP techniques are widely used in chatbots, sentiment analysis, and translation systems."""
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter
import string

# Convert to lowercase
text = text.lower()

# Tokenization
tokens = word_tokenize(text)
print("Tokens:\n", tokens)

# Remove punctuation and stopwords using a normal for loop
stop_words = set(stopwords.words('english'))
clean_tokens = []

for word in tokens:
    # Keep only alphabetic words (ignore punctuation/numbers)
    if word.isalpha():
        # Check if word is not a stopword
        if word not in stop_words:
            clean_tokens.append(word)

print("Clean Tokens (after removing stopwords and punctuation):")
print(clean_tokens)

print("\nClean Tokens (after removing stopwords and punctuation):\n", clean_tokens)

# Count frequency of each word
word_freq = Counter(clean_tokens)

print("\nWord Frequency:\n",word_freq)
print("\n",word_freq.most_common(10))
```
