## Named Entity Recognition using spaCy
### Question

1. Write a Python program to perform Named Entity Recognition (NER) using spaCy.
2. Load the English NLP model
3. Identify and display all named entities, their labels, and their positions in the text.
4. Print each entity with its corresponding category (like PERSON, ORG, GPE, DATE, etc.).

```python
text = """Elon Musk is the CEO of SpaceX and Tesla. 
He was born in South Africa and later moved to the United States. 
In 2021, SpaceX launched several Starlink satellites into orbit. 
Tesla’s headquarters are in California."""
```

```python
import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

# Display named entities
print("Named Entities, Phrases, and Labels:\n")
for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")
```
