## Sentiment Analysis using VADER
### Question

Perform sentiment analysis on the above reviews using the VADER sentiment analyzer.<br>
Display each sentence with its sentiment score and label it as Positive, Negative, or Neutral.

```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')

sentences = [
    "The movie was amazing and full of surprises.",
    "I hated the ending, it was too predictable.",
    "The soundtrack was great and the visuals were stunning.",
    "The story was boring and lacked creativity."
]

sid = SentimentIntensityAnalyzer()

for sentence in sentences:
    score = sid.polarity_scores(sentence)
    compound = score['compound']
    if compound > 0.05:
        sentiment = 'Positive'
    elif compound < -0.05:
        sentiment = 'Negative'
    else:
        sentiment = 'Neutral'
    print(f"{sentence}\nSentiment: {sentiment}, Score: {compound}\n")
```
