```python
# Tokenization, stopwords, stemming, lemmatization.

text = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence.
It helps computers understand, interpret, and respond to human language.
NLP techniques are widely used in chatbots, sentiment analysis, and translation systems."""

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter
import string

# Convert to lowercase
text = text.lower()

# Tokenization
tokens = word_tokenize(text)
print("Tokens:\n", tokens)

# Remove punctuation and stopwords using a normal for loop
stop_words = set(stopwords.words('english'))
clean_tokens = []

for word in tokens:
    # Keep only alphabetic words (ignore punctuation/numbers)
    if word.isalpha():
        # Check if word is not a stopword
        if word not in stop_words:
            clean_tokens.append(word)

print("Clean Tokens (after removing stopwords and punctuation):")
print(clean_tokens)

print("\nClean Tokens (after removing stopwords and punctuation):\n", clean_tokens)

# Count frequency of each word
word_freq = Counter(clean_tokens)

print("\nWord Frequency:\n",word_freq)
print("\n",word_freq.most_common(10))
```

```python
# Sentiment Analysis
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')

sentences = [
    "The movie was amazing and full of surprises.",
    "I hated the ending, it was too predictable.",
    "The soundtrack was great and the visuals were stunning.",
    "The story was boring and lacked creativity."
]

sid = SentimentIntensityAnalyzer()

for sentence in sentences:
    score = sid.polarity_scores(sentence)
    compound = score['compound']
    if compound > 0.05:
        sentiment = 'Positive'
    elif compound < -0.05:
        sentiment = 'Negative'
    else:
        sentiment = 'Neutral'
    print(f"{sentence}\nSentiment: {sentiment}, Score: {compound}\n")
```

```python
## Named Entity Recognition using spaCy

import spacy

text = """Elon Musk is the CEO of SpaceX and Tesla. 
He was born in South Africa and later moved to the United States. 
In 2021, SpaceX launched several Starlink satellites into orbit. 
Tesla’s headquarters are in California."""
# Load English model
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

# Display named entities
print("Named Entities, Phrases, and Labels:\n")
for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")
```
